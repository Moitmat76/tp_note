Version 2 : le calcul des moyennes des UE est réalisé avec des notes et des coefficients dans des listes.

```python
UE = int(input("Donnez-moi votre UE : "))

notes = []
coefficients = [10, 10, 20, 20]

for i in range(2):
    note = float(input("Donnez-moi la note de la ressource R{} : ".format(101 + i)))
    notes.append(note)

for i in range(2):
    note = float(input("Donnez-moi la note de la ressource SAE{} : ".format(11 + i)))
    notes.append(note)

somme = 0
somme_coeff = 0

for i in range(len(notes)):
    somme += notes[i] * coefficients[i]
    somme_coeff += coefficients[i]

moyenne = somme / somme_coeff

print("La moyenne de votre UE{} est de : {}".format(UE, moyenne))
```

```python
from tkinter import *
from tkinter import ttk
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image, ImageTk

# Création de la fenêtre
fen = Tk()
fen.title("Calculs des moyennes avec affichage graphique")
fen.geometry("1000x700")

# Création d'un cadre
fram = ttk.Frame(fen)
fram.grid(column=0, row=0, sticky="nw")

# Création des matières
matiere = [f"R{i}" for i in range(101, 116)] + [f"SAE{i}" for i in range(11, 17)]

# Création des listes de coefficients
coef1 = [10,10,7,7,0,5,0,6,0,5,4,2,5,5,0,20,20,0,0,0,7]
coef2 = [4,0,2,8,6,0,0,0,0,5,5,2,9,9,3,0,0,29,0,0,7]
coef3 = [4,0,2,0,0,5,15,6,4,5,5,2,0,0,3,0,0,0,20,20,7]

saisies = []

# Création des champs de saisie
for i, mat in enumerate(matiere):
    mat = ttk.Label(fram, text=mat)
    mat.grid(column=0, row=i+1, padx=10, pady=5)

    data = ttk.Entry(fram, width=10)
    data.grid(column=1, row=i+1)
    saisies.append(data)

# Création du graphique
fig, ax = plt.subplots(figsize=(5, 4))

canvas = FigureCanvasTkAgg(fig, master=fen)
canvas.get_tk_widget().grid(column=1, row=0, padx=10, pady=10)


def recup():
    notes = [float(data.get()) for data in saisies]

    # Calcul des moyennes
    moyenne1 = np.average(notes, weights=coef1)
    moyenne2 = np.average(notes, weights=coef2)
    moyenne3 = np.average(notes, weights=coef3)

    # Affichage des moyennes
    UE1 = ttk.Label(fram, text="Moyenne de l'UE1")
    UE1.grid(column=2, row=3, pady=10)

    moy1 = ttk.Label(fram, text=round(moyenne1, 3), relief=GROOVE)
    moy1.grid(column=3, row=3, pady=10)

    UE2 = ttk.Label(fram, text="Moyenne de l'UE2")
    UE2.grid(column=2, row=5, pady=10)

    moy2 = ttk.Label(fram, text=round(moyenne2, 3), relief=GROOVE)
    moy2.grid(column=3, row=5, pady=10)

    UE3 = ttk.Label(fram, text="Moyenne de l'UE3")
    UE3.grid(column=2, row=7, pady=10)

    moy3 = ttk.Label(fram, text=round(moyenne3, 3), relief=GROOVE)
    moy3.grid(column=3, row=7, pady=10)

    # Mise à jour du graphique
    ax.clear()

    UE = ["UE1", "UE2", "UE3"]
    moyennes = [moyenne1, moyenne2, moyenne3]

    ax.bar(UE, moyennes, color=["blue", "green", "red"])

    ax.set_title("Moyennes des UE")
    ax.set_ylabel("Notes")
    ax.set_ylim(0, 20)
    fig.savefig("moyenne_des_UE.jpg")
    canvas.draw()
    
# Bouton de validation
btn1 = Button(fram, text="Valider la saisie", command=recup)
btn1.grid(column=3, row=1, padx=15)

# Maintient la fenêtre ouverte
fen.mainloop()
```

```python
from tkinter import *
from tkinter import ttk
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image, ImageTk

# Création de la fenêtre
fen = Tk()
fen.title("Calculs des moyennes avec affichage graphique")
fen.geometry("1000x700")

# Création d'un cadre
fram = ttk.Frame(fen)
fram.grid(column=0, row=0, sticky="nw")

# Création des matières
matiere = [f"R{i}" for i in range(101, 116)] + [f"SAE{i}" for i in range(11, 17)]

# Création des listes de coefficients
coef1 = [10,10,7,7,0,5,0,6,0,5,4,2,5,5,0,20,20,0,0,0,7]
coef2 = [4,0,2,8,6,0,0,0,0,5,5,2,9,9,3,0,0,29,0,0,7]
coef3 = [4,0,2,0,0,5,15,6,4,5,5,2,0,0,3,0,0,0,20,20,7]

saisies = []

# Création des champs de saisie
for i, mat in enumerate(matiere):
    mat = ttk.Label(fram, text=mat)
    mat.grid(column=0, row=i+1, padx=10, pady=5)

    data = ttk.Entry(fram, width=10)
    data.grid(column=1, row=i+1)
    saisies.append(data)

# Création du graphique
fig, ax = plt.subplots(figsize=(5, 4))

canvas = FigureCanvasTkAgg(fig, master=fen)
canvas.get_tk_widget().grid(column=1, row=0, padx=10, pady=10)


def recup():
    notes = [float(data.get()) for data in saisies]

    # Calcul des moyennes
    moyenne1 = np.average(notes, weights=coef1)
    moyenne2 = np.average(notes, weights=coef2)
    moyenne3 = np.average(notes, weights=coef3)

    # Affichage des moyennes
    UE1 = ttk.Label(fram, text="Moyenne de l'UE1")
    UE1.grid(column=2, row=3, pady=10)

    moy1 = ttk.Label(fram, text=round(moyenne1, 3), relief=GROOVE)
    moy1.grid(column=3, row=3, pady=10)

    UE2 = ttk.Label(fram, text="Moyenne de l'UE2")
    UE2.grid(column=2, row=5, pady=10)

    moy2 = ttk.Label(fram, text=round(moyenne2, 3), relief=GROOVE)
    moy2.grid(column=3, row=5, pady=10)

    UE3 = ttk.Label(fram, text="Moyenne de l'UE3")
    UE3.grid(column=2, row=7, pady=10)

    moy3 = ttk.Label(fram, text=round(moyenne3, 3), relief=GROOVE)
    moy3.grid(column=3, row=7, pady=10)

    # Mise à jour du graphique


    #couleur du graph
    def code_couleur(moy):
        if moy >= 10:
            return "green"
        elif moy < 8: 
            return "red"
        else :
            return "orange"

    moyS1 = [moyenne1,moyenne2,moyenne3]
    colors = [code_couleur(moy) for moy in moyS1]
    ax.clear()
    ax.bar(["UE1", "UE2", "UE3"] ,moyS1, color=colors)

    ax.set_title("Moyennes des UE")
    ax.set_ylabel("Notes")
    ax.set_ylim(0, 20)
    fig.savefig("moyenne_des_UE.jpg")
    canvas.draw()
    
# Bouton de validation
btn1 = Button(fram, text="Valider la saisie", command=recup)
btn1.grid(column=3, row=1, padx=15)

# Maintient la fenêtre ouverte
fen.mainloop()
```
