Version 2 : le calcul des moyennes des UE est réalisé avec des notes et des coefficients dans des listes.

```python
UE = int(input("Donnez-moi votre UE : "))

notes = []
coefficients = [10, 10, 20, 20]

for i in range(2):
    note = float(input("Donnez-moi la note de la ressource R{} : ".format(101 + i)))
    notes.append(note)

for i in range(2):
    note = float(input("Donnez-moi la note de la ressource SAE{} : ".format(11 + i)))
    notes.append(note)

somme = 0
somme_coeff = 0

for i in range(len(notes)):
    somme += notes[i] * coefficients[i]
    somme_coeff += coefficients[i]

moyenne = somme / somme_coeff

print("La moyenne de votre UE{} est de : {}".format(UE, moyenne))
```

```python
UE = int(input("Donnez-moi votre UE : "))

coefficients = {
    1: {
        "R101": 10, "R102": 10, "R103": 7, "R104": 7,
        "R106": 5, "R108": 6, "R110": 5, "R111": 4,
        "R112": 2, "R113": 5, "R114": 5,
        "SAE11": 20, "SAE12": 20, "SAE16": 7
    },

    2: {
        "R101": 4, "R103": 2, "R104": 8, "R105": 6,
        "R110": 5, "R111": 5, "R112": 2, "R113": 9,
        "R114": 9, "R115": 3,
        "SAE13": 29, "SAE16": 7
    },

    3: {
        "R101": 4, "R103": 2, "R106": 5, "R107": 15,
        "R108": 6, "R109": 4, "R110": 5, "R111": 5,
        "R112": 2, "R115": 3,
        "SAE14": 20, "SAE15": 20, "SAE16": 7
    }
}

notes = []
coeffs = coefficients[UE]

for ressource, coeff in coeffs.items():
    note = float(input("Donnez-moi la note de {} : ".format(ressource)))
    notes.append(note)

somme = 0
somme_coeff = 0

for i in range(len(notes)):
    somme += notes[i] * list(coeffs.values())[i]
    somme_coeff += list(coeffs.values())[i]

moyenne = somme / somme_coeff

print("La moyenne de votre UE{} est de : {:.2f}".format(UE, moyenne))
```
